```python
budgets = pd.read_csv("../data/budgets_ecuries.csv")
print("Nombre de lignes :", len(budgets))
budgets.head()
```

```python
ecuries_kaggle = sorted(classement_final["name"].unique())
print("=== Écuries dans le dataset Kaggle ===")
for nom in ecuries_kaggle:
    print(" -", nom)

ecuries_budgets = sorted(budgets["ecurie"].unique())
print("\n=== Écuries dans nos budgets ===")
for nom in ecuries_budgets:
    print(" -", nom)
```

```python
mapping_ecuries = {
    "Alfa Romeo":      "Kick Sauber",   # Sauber renommé Alfa Romeo 2019-2023
    "AlphaTauri":      "Racing Bulls",  # Toro Rosso renommé AlphaTauri 2020-2023
    "Alpine F1 Team":  "Alpine",
    "Aston Martin":    "Aston Martin",
    "Caterham":        "Caterham",
    "Ferrari":         "Ferrari",
    "Force India":     "Aston Martin",  # Devient Racing Point puis Aston Martin
    "Haas F1 Team":    "Haas",
    "Lotus F1":        "Alpine",        # Devient Renault puis Alpine
    "Manor Marussia":  "Manor",
    "Marussia":        "Marussia",
    "McLaren":         "McLaren",
    "Mercedes":        "Mercedes",
    "RB F1 Team":      "Racing Bulls",  # AlphaTauri renommé RB en 2024
    "Racing Point":    "Aston Martin",
    "Red Bull":        "Red Bull",
    "Renault":         "Alpine",
    "Sauber":          "Kick Sauber",
    "Toro Rosso":      "Racing Bulls",
    "Williams":        "Williams",
}

classement_final["ecurie"] = classement_final["name"].map(mapping_ecuries)

non_matche = classement_final[classement_final["ecurie"].isna()]
print("Lignes sans correspondance :", len(non_matche))
if len(non_matche) > 0:
    print(non_matche[["year", "name"]])
    
```

```python

budgets_renamed = budgets.rename(columns={"saison": "year"})

analyse = classement_final.merge(
    budgets_renamed,
    on=["year", "ecurie"],
    how="inner"  
)

print("Nombre de lignes dans ton dataset analytique :", len(analyse))
print("\nColonnes disponibles :")
print(analyse.columns.tolist())
analyse.head()
```

```python

analyse["roi_points_par_musd"] = analyse["points"] / analyse["budget_total_estime_musd"]

print("=== TOP 10 MEILLEURS ROI (points par million $) ===\n")
top_roi = analyse[["year", "ecurie", "position", "points", "budget_total_estime_musd", "roi_points_par_musd"]] \
    .sort_values("roi_points_par_musd", ascending=False) \
    .head(10)

top_roi
```

```python
analyse["periode"] = analyse["year"].apply(
    lambda y: "Cost cap (2021+)" if y >= 2021 else "Pré cost cap (2014-2020)"
)

print(analyse["periode"].value_counts())
analyse[["year", "ecurie", "periode"]].head(5)
```

```python
roi_moyen = analyse.groupby("ecurie")["roi_points_par_musd"].mean().sort_values(ascending=False)
print("=== ROI MOYEN PAR ÉCURIE (2014-2024) ===\n")
print(roi_moyen.round(2))
```

```python
colonnes_export = [
    "year", "ecurie", "nom_historique", "position", "points", "wins",
    "budget_plafonne_musd", "budget_total_estime_musd", "prize_money_musd",
    "roi_points_par_musd", "periode", "fiabilite"
]

dataset_final = analyse[colonnes_export].copy()
dataset_final = dataset_final.sort_values(["year", "position"])

dataset_final.to_csv("../data/dataset_analytique.csv", index=False)
print("Dataset sauvegardé dans data/dataset_analytique.csv")
print("Nombre de lignes :", len(dataset_final))
dataset_final.head(15)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 6))

for periode, group in analyse.groupby("periode"):
    ax.scatter(group["budget_total_estime_musd"], group["points"], 
               label=periode, alpha=0.6, s=80)

ax.set_xlabel("Budget total estimé (M$)")
ax.set_ylabel("Points au championnat constructeurs")
ax.set_title("Budget vs Performance des écuries F1 (2014-2024)")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
